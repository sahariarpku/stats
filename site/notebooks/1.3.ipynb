## The question

A teacher has five quiz scores: 72, 85, 90, 68 and 95. The average is 82, but she wants one number for how spread out the scores are. She also wonders why her software offers two different standard deviations, and which one to trust.

## Step 1: By hand: the six steps

Follow the lesson: find the mean, the deviations, square them, add them up (the sum of squares), divide by n - 1 and take the square root. The raw deviations always add to zero.

````python
import numpy as np  # Load NumPy, the library for lists of numbers, as np.

score = np.array([72, 85, 90, 68, 95])  # Store the five quiz scores in an array so we can do arithmetic on all of them at once.

mean = score.sum() / len(score)  # Step 1, the mean: total divided by the count. This is the centre the spread is measured from.
dev = score - mean  # Step 2, the deviations: how far each score sits from the mean. NumPy subtracts the mean from all five at once.
ss = (dev ** 2).sum()  # Steps 3 and 4: square each deviation (this removes minus signs) and add them up. This is the sum of squares.
var = ss / (len(score) - 1)  # Step 5, the variance: divide by n - 1, not n, because a sample sits closer to its own mean than to the true mean, so dividing by n would under-estimate the spread.

print('mean =', mean, ' deviations =', dev, ' sum =', dev.sum())  # Print the mean, the deviations and their sum. The sum is zero, which is why we cannot average raw deviations.
print('sum of squares =', ss)  # Print the sum of squares.
print(f'sample variance = {var:.1f}, sample sd = {np.sqrt(var):.2f}')  # Step 6: the square root brings the variance back to score units. That gives the standard deviation.
````

**What this means.** The mean is 82. The deviations -10, 3, 8, -14 and 13 add to 0, so they must be squared. The sum of squares is 538, the sample variance is 538 / 4 = 134.5, and the sample standard deviation is 11.60 points: a typical score lies about 11.6 points from 82.

## Step 2: The same with NumPy: ddof 0 or 1

NumPy lets you choose the divisor with ddof (delta degrees of freedom). ddof=1 divides by n - 1 (a sample), ddof=0 divides by n (a whole population).

````python
print(f'sample     variance = {score.var(ddof=1):.1f}, sd = {score.std(ddof=1):.2f}')  # ddof=1 divides by n - 1. Use it when the scores are a sample from a bigger group.
print(f'population variance = {score.var(ddof=0):.1f}, sd = {score.std(ddof=0):.2f}')  # ddof=0 divides by n. Use it only when you hold every member of the group.
````

**What this means.** The sample version gives variance 134.5 and SD 11.60, matching our hand work. The population version gives variance 107.6 and SD 10.37. The sample version is bigger on purpose, to correct for samples looking less spread out than the whole group. With five scores the difference is large, and with hundreds it almost vanishes.

## Step 3: A trap: NumPy and pandas differ

np.std defaults to ddof=0 but pandas .std() defaults to ddof=1. Always say which one you mean. R, SPSS and Stata all use the sample version (n - 1) by default.

````python
import pandas as pd  # Load pandas, Python's tool for tables and columns of data, as pd.

s = pd.Series(score)  # Wrap the scores in a pandas Series (a column of data) to see what pandas does by default.
print('np.std(score)     =', round(np.std(score), 2), '(population)')  # np.std defaults to ddof=0, the population version.
print('pd.Series.std()   =', round(s.std(), 2), '(sample)')  # pandas .std() defaults to ddof=1, the sample version.
````

**What this means.** For the same five numbers np.std gives 10.37 and pandas gives 11.6, because their defaults differ. Neither is wrong, but you must know which you used and say so. R, SPSS and Stata all use the sample version (n - 1) by default.

## Step 4: Same mean, different spread

Dataset A is perfectly steady and B swings widely, yet both have mean 50. The standard deviation tells them apart. Here B is treated as a whole population (divide by N).

````python
A = np.array([50, 50, 50, 50])  # Dataset A: four identical values. Nothing varies.
B = np.array([10, 30, 70, 90])  # Dataset B: four values far from 50 on both sides, but with the same mean as A.

print('means:', A.mean(), B.mean())  # Print both means to show they are equal.
print('sd of A:', A.std(ddof=0))  # The SD of A, treating it as a complete group (ddof=0).
print(f'sd of B: {B.std(ddof=0):.2f}  (variance {B.var(ddof=0):.0f})')  # The SD and variance of B. :.0f shows the variance as a whole number.
````

**What this means.** Both datasets have a mean of 50, but A's SD is 0.0 and B's is 31.62 (variance 1000). The mean cannot tell them apart and the SD can. An SD of exactly 0 happens only when every value is identical.

## So, what is the answer?

The five scores average 82 and their sample standard deviation is 11.6, so a typical score is about 11.6 points from the mean (variance 134.5, in squared points). Use n - 1 for a sample, and n only when you hold the entire group. Caution: with five numbers the SD is a rough estimate, and one extreme score would change it a lot.

## Now you try

A factory has two machines that fill 500 ml bottles. The quality manager measures six bottles from each machine. Both machines average 500 ml, but she suspects one is much less steady than the other.

**Data:** machine_a = [499, 501, 500, 502, 498, 500]; machine_b = [490, 510, 495, 505, 500, 500]

1. Find the mean of each machine.
2. Find the sample standard deviation of each machine.
3. Say which machine is steadier, and by how many times.
4. Compute the population SDs too, and say which version suits this problem.

*Hint:* Use np.mean and np.std with ddof=1. A sample of six bottles is not every bottle the machine will ever fill.

````python
# Your turn: write your code here
````

### Solution (try first, then peek)

````python
import numpy as np

a = np.array([499, 501, 500, 502, 498, 500])
b = np.array([490, 510, 495, 505, 500, 500])
for name, x in [('A', a), ('B', b)]:
    print(f'machine {name}: mean = {x.mean():.1f}, sample sd = {x.std(ddof=1):.2f}, '
          f'population sd = {x.std(ddof=0):.2f}')
print(f'variance of B (sample) = {b.var(ddof=1):.1f}')
````

**Answer.** Both machines average 500.0 ml. Machine A has a sample SD of 1.41 ml and machine B 7.07 ml, five times larger, so B is far less steady even though its mean is perfect. The population SDs (1.29 and 6.45) are smaller. Use the sample SD here, because six bottles are only a sample of what each machine fills.

## The same question in R, SPSS and Stata

**R**

```r
score <- c(72, 85, 90, 68, 95)  # Store the five quiz scores in a vector.
mean(score)  # mean() gives the centre of the scores.
var(score)  # var() gives the sample variance, using the divisor n - 1.
sd(score)  # sd() gives the sample standard deviation, the square root of that variance.
# population version: divide by n instead of n - 1  # A comment: R ignores everything after #. We use it to note that the next lines change the divisor.
var(score) * (length(score) - 1) / length(score)  # R has no built-in population variance, so we convert: undo the n - 1 divisor and divide by n instead.
sqrt(var(score) * (length(score) - 1) / length(score))  # The square root of that population variance is the population standard deviation.
```

How to read it: R prints 82 (mean), 134.5 (variance) and 11.59741 (SD). These are the sample versions. The last two lines print the population variance 107.6 and the population SD 10.37.
