## 1. By hand: the six steps

Follow the lesson: find the mean, the deviations, square them, add them up (the sum of squares), divide by n - 1 and take the square root. The raw deviations always add to zero.

````python
import numpy as np

score = np.array([72, 85, 90, 68, 95])

mean = score.sum() / len(score)
dev = score - mean
ss = (dev ** 2).sum()
var = ss / (len(score) - 1)

print('mean =', mean, ' deviations =', dev, ' sum =', dev.sum())
print('sum of squares =', ss)
print(f'sample variance = {var:.1f}, sample sd = {np.sqrt(var):.2f}')
````

## 2. The same with NumPy: ddof 0 or 1

NumPy lets you choose the divisor with ddof (delta degrees of freedom). ddof=1 divides by n - 1 (a sample), ddof=0 divides by n (a whole population).

````python
print(f'sample     variance = {score.var(ddof=1):.1f}, sd = {score.std(ddof=1):.2f}')
print(f'population variance = {score.var(ddof=0):.1f}, sd = {score.std(ddof=0):.2f}')
````

## 3. A trap: NumPy and pandas differ

np.std defaults to ddof=0 but pandas .std() defaults to ddof=1. Always say which one you mean. R, SPSS and Stata all use the sample version (n - 1) by default.

````python
import pandas as pd

s = pd.Series(score)
print('np.std(score)     =', round(np.std(score), 2), '(population)')
print('pd.Series.std()   =', round(s.std(), 2), '(sample)')
````

## 4. Same mean, different spread

Dataset A is perfectly steady and B swings widely, yet both have mean 50. The standard deviation tells them apart. Here B is treated as a whole population (divide by N).

````python
A = np.array([50, 50, 50, 50])
B = np.array([10, 30, 70, 90])

print('means:', A.mean(), B.mean())
print('sd of A:', A.std(ddof=0))
print(f'sd of B: {B.std(ddof=0):.2f}  (variance {B.var(ddof=0):.0f})')
````

## The same idea in R, SPSS and Stata

**R**

```r

mean(score)
var(score)
sd(score)
# population version: divide by n instead of n - 1
var(score) * (length(score) - 1) / length(score)
sqrt(var(score) * (length(score) - 1) / length(score))
```
