## Step 1: Factorials

n! is the number of ways to line up n different things. Python's math module has factorial built in. By convention 0! equals 1.

````python
import math  # Load Python's math module, which has counting functions built in.

for n in [0, 1, 4, 5, 8]:  # Loop over a few values of n to see how fast factorials grow.
    print(f'{n}! = {math.factorial(n)}')  # Print n! for each one. The factorial n! is the number of ways to line up n different things.
````

**What this means.** 0! = 1 (by convention, one way to arrange nothing), 4! = 24, 5! = 120 and 8! = 40320. Factorials grow very fast, which is why counting by hand quickly becomes impossible and we need formulas.

## Step 2: Permutations and combinations

math.perm(n, r) counts ordered selections (a podium of 3 from 8 runners). math.comb(n, r) counts unordered ones (a 3-person committee from 10). Dividing the permutations by r! gives the combinations.

````python
print('podiums, perm(8, 3)       =', math.perm(8, 3))  # Count ordered selections: a podium of 3 from 8 runners. perm(n, r) is used when the order of the picks matters.
print('committees, comb(10, 3)   =', math.comb(10, 3))  # Count unordered selections: a 3-person committee from 10. comb(n, r) is used when only who is picked matters, not the order.
print('perm(10, 3) / 3!          =', math.perm(10, 3) // math.factorial(3))  # Divide the permutations by 3! (the ways to reorder 3 people). This shows the link between the two counts.
print('poker hands, comb(52, 5)  =', math.comb(52, 5))  # Count 5-card poker hands. A hand is a group, so order does not matter.
````

**What this means.** There are 336 podiums but only 120 committees. The third line shows why: perm(10, 3) / 3! is also 120, so each committee of 3 appears in 6 different orders. A poker hand can be chosen in 2,598,960 ways.

## Step 3: List them to confirm

For small cases we can write out every selection with itertools and count it. Choosing 3 of 5 people gives 60 ordered lists but only 10 groups, a ratio of 3! = 6.

````python
from itertools import permutations, combinations  # Load permutations and combinations, which list the actual selections and not just count them.

people = ['A', 'B', 'C', 'D', 'E']  # Name five people so we can list real selections.
ordered = list(permutations(people, 3))  # List every ordered way to pick 3 of the 5. Each item is a different line-up.
groups = list(combinations(people, 3))  # List every group of 3 of the 5, ignoring order.

print('ordered lists:', len(ordered), ' formula:', math.perm(5, 3))  # Compare the number of ordered lists with the formula. They must agree.
print('groups:', len(groups), ' formula:', math.comb(5, 3))  # Compare the number of groups with the formula, and show the groups agree too.
print('first groups:', [''.join(g) for g in groups[:4]])  # Show the first four groups to see what a combination looks like.
````

**What this means.** Listing gives 60 ordered lists and 10 groups, the same as the formulas perm(5, 3) = 60 and comb(5, 3) = 10. The ratio is 6 = 3!. Checking a formula against a small list is how you gain trust in it before using it on big numbers.

## Step 4: From counting to probability

One lottery ticket wins out of every possible set of 6 numbers from 49. The chance is 1 divided by the number of tickets. The all-hearts hand is (ways to pick 5 hearts) divided by (all hands).

````python
tickets = math.comb(49, 6)  # Count all possible lottery tickets: ways to choose 6 numbers from 49.
print(f'tickets = {tickets:,}')  # Print the count with thousands separators.
print(f'P(jackpot) = {1 / tickets:.2e}')  # One ticket wins, so the chance is 1 divided by the number of tickets. The e notation means a very small number.

hearts = math.comb(13, 5) / math.comb(52, 5)  # Count the hands made of 5 hearts and divide by all hands to get the probability.
print(f'P(5 hearts) = {math.comb(13, 5)} / {math.comb(52, 5):,} = {hearts:.4f}')  # Print the counts and the probability so every piece is visible.
````

**What this means.** There are 13,983,816 tickets, so the jackpot chance is 7.15e-08 (about 1 in 14 million). The chance of a 5-card hand with all hearts is 1287 / 2,598,960 = 0.0005. Both are favourable counts divided by all counts, which is the basic probability rule.

## So, what is the answer?

Picking a captain, vice-captain and treasurer from 9 members can be done in 504 ways because order matters, and picking a 3-person team in 84 ways because it does not. The lottery jackpot has a 1 in 13,983,816 chance. The caution: the hard part is deciding whether order matters, since the wrong choice multiplies the count by 3! or r!.

**Data:** Members: 9. Offices: 3 distinct roles. Delegation: 3 people. Raffle: choose 5 numbers from 30.

2. Divide the first count by the second. What does the ratio mean?
3. What is the chance of winning the raffle with one ticket?
4. Ana and Ben are two of the 9. What is the chance both are in the delegation?

*Hint:* Offices have an order, so use math.perm; a delegation has none, so use math.comb. For Ana and Ben, count the delegations that already contain both: only the third member is free.

### Solution (try first, then peek)

````python
import math

captain_vice_treasurer = math.perm(9, 3)
delegation = math.comb(9, 3)
print('officer line-ups:', captain_vice_treasurer)
print('delegations     :', delegation)
print('ratio           :', captain_vice_treasurer // delegation, '= 3!')

draw = math.comb(30, 5)
print(f'raffle draws: {draw:,}, P(win) = {1 / draw:.2e}')
print(f'P(Ana and Ben both on the delegation) = {math.comb(7, 1)}/{delegation} = {math.comb(7, 1) / delegation:.4f}')
````

**Answer.** There are perm(9, 3) = 504 ways to fill the offices and comb(9, 3) = 84 delegations. The ratio is 6, which is 3!, the number of ways to reorder the same 3 people. The raffle has 142,506 draws, so one ticket wins with probability 7.02e-06. With Ana and Ben fixed, 7 delegations remain, so the chance is 7/84 = 0.0833.

