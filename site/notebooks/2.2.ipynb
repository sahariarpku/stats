## 1. Factorials

n! is the number of ways to line up n different things. Python's math module has factorial built in. By convention 0! equals 1.

````python
import math

for n in [0, 1, 4, 5, 8]:
    print(f'{n}! = {math.factorial(n)}')
````

## 2. Permutations and combinations

math.perm(n, r) counts ordered selections (a podium of 3 from 8 runners). math.comb(n, r) counts unordered ones (a 3-person committee from 10). Dividing the permutations by r! gives the combinations.

````python
print('podiums, perm(8, 3)       =', math.perm(8, 3))
print('committees, comb(10, 3)   =', math.comb(10, 3))
print('perm(10, 3) / 3!          =', math.perm(10, 3) // math.factorial(3))
print('poker hands, comb(52, 5)  =', math.comb(52, 5))
````

## 3. List them to confirm

For small cases we can write out every selection with itertools and count it. Choosing 3 of 5 people gives 60 ordered lists but only 10 groups, a ratio of 3! = 6.

````python
from itertools import permutations, combinations

people = ['A', 'B', 'C', 'D', 'E']
ordered = list(permutations(people, 3))
groups = list(combinations(people, 3))

print('ordered lists:', len(ordered), ' formula:', math.perm(5, 3))
print('groups:', len(groups), ' formula:', math.comb(5, 3))
print('first groups:', [''.join(g) for g in groups[:4]])
````

## 4. From counting to probability

One lottery ticket wins out of every possible set of 6 numbers from 49. The chance is 1 divided by the number of tickets. The all-hearts hand is (ways to pick 5 hearts) divided by (all hands).

````python
tickets = math.comb(49, 6)
print(f'tickets = {tickets:,}')
print(f'P(jackpot) = {1 / tickets:.2e}')

hearts = math.comb(13, 5) / math.comb(52, 5)
print(f'P(5 hearts) = {math.comb(13, 5)} / {math.comb(52, 5):,} = {hearts:.4f}')
````
