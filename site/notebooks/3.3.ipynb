## The question

An emergency room averages 4 patient arrivals per hour. The manager asks: how likely is exactly 2 arrivals in an hour (a quiet hour), and how likely is a rush of 8 or more? Can the Poisson distribution, which needs only the average rate, answer this?

## Step 1: Set the average rate

The Poisson distribution only needs one number: the average count per interval, called lambda. The ER averages 4 arrivals an hour, so we build a Poisson distribution with that rate.

````python
from scipy import stats  # Load the statistics tools from SciPy.

lam = 4  # Store the average rate. The Poisson distribution is described by this one number, called lambda.
er = stats.poisson(mu=lam)  # Build a Poisson distribution with that rate for the emergency room.

print('mean     =', er.mean())  # Print its mean.
print('variance =', er.var())  # Print its variance.
````

**What this means.** The mean is 4.0 and the variance is 4.0. For a Poisson distribution the mean and variance are equal, and seeing them match is the quick check that this model fits count data of this kind.

## Step 2: Exactly 2 arrivals

pmf gives the probability of one exact count. Notice the mean and the variance are both 4: that equality is the Poisson's fingerprint.

````python
p2 = er.pmf(2)  # Ask for the probability of exactly 2 arrivals with pmf (probability mass function).

print(f'P(X = 2) = {p2:.4f}')  # Print it with four decimals.
````

**What this means.** P(X = 2) = 0.1465, so about 14.7% of hours bring exactly 2 arrivals. That is common but not the most likely count: 3 and 4 are both likelier.

## Step 3: A rush of 8 or more

'8 or more' is everything except 0 to 7. The survival function sf(7) is P(X > 7), which is exactly that, so it is shorter and more accurate than adding eight terms.

````python
rush = er.sf(7)  # The survival function sf(7) is the chance of more than 7, which means 8 or more.

print(f'P(X >= 8) = {rush:.4f}')  # Print the chance of a rush.
print(f'about 1 hour in {1 / rush:.0f}')  # Convert it to 'about 1 hour in N' to make it easier to picture.
````

**What this means.** P(X >= 8) = 0.0511, about 5%, or roughly 1 hour in 20. A rush like this is unusual but not rare, so the ER should expect it a few times a week.

## Step 4: See the whole distribution

A table of probabilities shows where the mass sits. Most hours bring 2 to 6 patients, and the long right tail is the rare rush.

````python
import pandas as pd  # Load pandas for the table.

k = list(range(0, 11))  # List the counts 0 to 10.
table = pd.DataFrame({'k': k, 'P(X = k)': er.pmf(k).round(4)})  # Build a table of counts next to their probabilities, rounded to four decimals.

print(table.to_string(index=False))  # Print the table without the row numbers.
````

**What this means.** The table peaks at 3 and 4 arrivals (0.1954 each), and most of the probability sits between 2 and 6. The tail thins out: 8 arrivals has only 0.0298, and 10 only 0.0053.

## So, what is the answer?

With an average of 4 arrivals an hour, exactly 2 arrivals happens 14.7% of the time and a rush of 8 or more happens 5.1% of the time, roughly 1 hour in 20. The caution: the Poisson model assumes arrivals are independent and the rate is constant, which may not hold during busy periods such as Friday nights.

### Solution (try first, then peek)

````python
from scipy import stats

desk = stats.poisson(mu=6)
print('mean, variance:', desk.mean(), desk.var())
print(f'P(X = 4)   = {desk.pmf(4):.4f}')
print(f'P(X <= 2)  = {desk.cdf(2):.4f}')
print(f'P(X >= 10) = {desk.sf(9):.4f}')
print(f'about 1 window in {1 / desk.sf(9):.0f}')

calls = stats.poisson(6).rvs(10_000, random_state=1)
print(f'simulated mean {calls.mean():.3f}, variance {calls.var():.3f}')
````

**Answer.** The mean and variance are both 6.0. P(X = 4) = 0.1339, P(X <= 2) = 0.0620 and P(X >= 10) = 0.0839, so an overwhelming window comes about 1 in 12. The 10,000 simulated windows had a mean of 5.978 and a variance of 5.901, both near 6, which is the Poisson fingerprint of equal mean and variance.

## The same question in R, SPSS and Stata

```r
1 - ppois(7, lambda = 4)  # ppois gives P(X <= 7); 1 minus it is P(X >= 8).
ppois(7, lambda = 4, lower.tail = FALSE)  # The same answer using lower.tail = FALSE for the upper tail directly.
```

How to read it: The first number is 0.1465 for exactly 2. The next two lines both give 0.0511, the chance of 8 or more, found two ways. They match Python.
