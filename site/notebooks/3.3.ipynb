## 1. Set the average rate

The Poisson distribution only needs one number: the average count per interval, called lambda. The ER averages 4 arrivals an hour, so we build a Poisson distribution with that rate.

````python
from scipy import stats

lam = 4
er = stats.poisson(mu=lam)

print('mean     =', er.mean())
print('variance =', er.var())
````

## 2. Exactly 2 arrivals

pmf gives the probability of one exact count. Notice the mean and the variance are both 4: that equality is the Poisson's fingerprint.

````python
p2 = er.pmf(2)

print(f'P(X = 2) = {p2:.4f}')
````

## 3. A rush of 8 or more

'8 or more' is everything except 0 to 7. The survival function sf(7) is P(X > 7), which is exactly that, so it is shorter and more accurate than adding eight terms.

````python
rush = er.sf(7)

print(f'P(X >= 8) = {rush:.4f}')
print(f'about 1 hour in {1 / rush:.0f}')
````

## 4. See the whole distribution

A table of probabilities shows where the mass sits. Most hours bring 2 to 6 patients, and the long right tail is the rare rush.

````python
import pandas as pd

k = list(range(0, 11))
table = pd.DataFrame({'k': k, 'P(X = k)': er.pmf(k).round(4)})

print(table.to_string(index=False))
````
