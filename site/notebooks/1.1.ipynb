## The question

A coffee shop with nine employees wants to quote an 'average salary' in a job advert. When the owner, who earns 200 thousand dollars, joins the staff list, the average jumps by 15 thousand even though nobody got a raise. Which number should the advert use: the mean, the median or the mode?

## Step 1: Mean and median of nine salaries

The lesson's nine coffee shop salaries, in thousands of dollars. The mean is the balance point and the median is the middle value of the sorted list.

````python
import numpy as np  # Load NumPy, the library for lists of numbers, as np.
import statistics  # Load the statistics module from Python's standard library; it has multimode, which we need for the mode.

salary = np.array([38, 42, 45, 47, 50, 52, 55, 58, 60])  # Store the nine salaries (in thousands of dollars) in an array, so one command works on all of them.

print(f'mean   = {salary.mean():.1f}')  # The mean (sum divided by count) is the balance point of the salaries. :.1f rounds it to one decimal.
print(f'median = {np.median(salary):.1f}')  # np.median sorts the values and takes the middle one, the 5th of 9: half earn less and half earn more.
````

**What this means.** The mean is 49.7 and the median 50.0. They are almost identical, which tells us the salaries are spread evenly around the middle. Either number describes a typical employee: about 50 thousand dollars.

## Step 2: The owner joins the staff list

Add the owner's 200 and compare. Nobody else got a raise, yet the mean jumps by 15 while the median moves by 1. The median only looks at position in the sorted list.

````python
with_owner = np.append(salary, 200)  # np.append returns a new array with the owner's 200 added, and leaves salary unchanged so we can compare before and after.

print(f'mean   {salary.mean():.1f} -> {with_owner.mean():.1f}')  # Print the mean before and after the owner joins.
print(f'median {np.median(salary):.1f} -> {np.median(with_owner):.1f}')  # Print the median before and after, in the same format so the two lines can be compared.
print(f'mean moved by {with_owner.mean() - salary.mean():.1f}, median by {np.median(with_owner) - np.median(salary):.1f}')  # Print how far each one moved, which is the number that answers 'who reacts to one big value?'.
````

**What this means.** The mean goes from 49.7 to 64.7, a jump of 15.0, although nobody else got a raise. The median goes from 50.0 to 51.0, a move of just 1.0. The mean uses every value, so one huge salary drags it. The median only looks at the middle position.

## Step 3: When the mean describes nobody

The lesson's small company: six salaries in thousands of dollars, with the founder at 800. The mean of 173.5 matches no employee. The median, 49.5, is a fair picture of a typical one.

````python
company = np.array([42, 45, 48, 51, 55, 800])  # A second company: six salaries in thousands, with a founder at 800.

print(f'mean   = {company.mean():.1f}')  # Print the mean, which adds in the founder's 800.
print(f'median = {np.median(company):.1f}')  # Print the median, which only needs the middle of the sorted list.
````

**What this means.** The mean is 173.5, higher than every employee's pay except the founder's, so it describes nobody. The median is 49.5, in the middle of what most staff earn. When one value is extreme, the median is the honest summary.

## Step 4: The mode: the most common value

statistics.multimode returns every value tied for most frequent, so it also shows when there are two modes. It is the only one of the three that works for categories such as colours.

````python
shoes = [7, 8, 8, 9, 9, 9, 10]  # Shoe sizes: a list of numbers where one size is the most popular.
ties = [1, 2, 2, 3, 3]  # A small list in which two values tie for the most frequent.
colours = ['red', 'blue', 'blue', 'green']  # A list of words. Mean and median do not exist for names, but the mode still does.

print('shoe sizes mode :', statistics.multimode(shoes))  # multimode returns a list of every most-frequent value, so a tie shows up instead of being hidden.
print('tied example    :', statistics.multimode(ties))  # Run it on the tied example to see two modes.
print('colours mode    :', statistics.multimode(colours))  # Run it on the colours to see that the mode works for categories.
````

**What this means.** Size 9 is the most common shoe size, appearing three times. In 1, 2, 2, 3, 3 the values 2 and 3 tie, so there are two modes. For the colours the mode is 'blue'. The mode is the only one of the three averages that works for categories.

## So, what is the answer?

For the job advert use the median: about 50 thousand dollars before the owner joins and 51 after, while the mean jumps from 49.7 to 64.7 because of one salary. The mean is fine when values are spread evenly, the median is safer when a few values are extreme, and the mode is the only choice for categories. Caution: one number always hides the spread.

An estate agent lists the sale prices (in thousands of dollars) of nine ordinary homes on Maple Street. Then a mansion on the same street sells for 1,450. She wants to quote the 'typical price' of a home on the street.

**Data:** street = [212, 225, 230, 238, 245, 251, 260, 268, 275]; mansion = 1450

1. Find the mean and median of the nine ordinary homes.
2. Add the mansion and recompute both.
3. Say how far each one moved, and which one you would quote.

*Hint:* Use np.append to add the mansion to a new array, then compare mean() and np.median() before and after.

### Solution (try first, then peek)

````python
import numpy as np

street = np.array([212, 225, 230, 238, 245, 251, 260, 268, 275])
with_mansion = np.append(street, 1450)
print(f'street:  mean = {street.mean():.1f}, median = {np.median(street):.1f}')
print(f'+mansion: mean = {with_mansion.mean():.1f}, median = {np.median(with_mansion):.1f}')
print(f'mean rose by {with_mansion.mean() - street.mean():.1f}, median by {np.median(with_mansion) - np.median(street):.1f}')
````

**Answer.** Without the mansion the mean is 244.9 and the median 245.0. With it the mean leaps to 365.4 (up 120.5) while the median is 248.0 (up just 3.0). Nine of ten homes cost under 280, so 365.4 describes no home on the street. Quote the median, about 248 thousand dollars.

## The same question in R, SPSS and Stata

**R**

```r
salary <- c(38, 42, 45, 47, 50, 52, 55, 58, 60)  # Store the nine salaries in a vector.
mean(salary); median(salary)  # Two commands on one line, separated by a semicolon. R prints the mean and then the median.
with_owner <- c(salary, 200)  # Add the owner's 200 to the end of the vector, making a new vector called with_owner.
mean(with_owner); median(with_owner)  # The mean and median again, now with the owner included.
shoes <- c(7, 8, 8, 9, 9, 9, 10)  # Store the shoe sizes in a vector.
table(shoes)  # table() counts how often each shoe size occurs.
names(which.max(table(shoes)))  # which.max finds the biggest count and names() gives the size it belongs to. R has no ready-made function for the mode.
```

How to read it: The first line prints 49.67 and 50. After the owner joins it prints 64.7 and 51: the mean moved by 15, the median by 1. table(shoes) shows size 9 appearing 3 times and the last line returns "9", the mode.

**SPSS** (Analyze > Descriptive Statistics > Frequencies > Statistics)

```
DATA LIST FREE / salary.
BEGIN DATA
38 42 45 47 50 52 55 58 60 200
END DATA.
FREQUENCIES VARIABLES=salary
  /FORMAT=NOTABLE
  /STATISTICS=MEAN MEDIAN MODE.
TEMPORARY.
SELECT IF (salary < 100).
FREQUENCIES VARIABLES=salary
  /FORMAT=NOTABLE
  /STATISTICS=MEAN MEDIAN MODE.
```

How to read it: In the first Statistics table (with the owner) the Mean is 64.70 and the Median 51.00. In the second (owner left out) the Mean is 49.67 and the Median 50.00. The Mode row shows 38, the smallest value, with a footnote: every value is different, so several modes exist.

**Stata**

```

```

How to read it: With the owner, the Mean is 64.7 and the 50% row (the median) is 51. Without the owner the Mean is 49.67 and the median 50. tabulate shoe shows size 9 with a frequency of 3, the largest, so the mode is 9.
