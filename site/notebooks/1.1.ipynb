## 1. Mean and median of nine salaries

The lesson's nine coffee shop salaries, in thousands of dollars. The mean is the balance point and the median is the middle value of the sorted list.

````python
import numpy as np
import statistics

salary = np.array([38, 42, 45, 47, 50, 52, 55, 58, 60])

print(f'mean   = {salary.mean():.1f}')
print(f'median = {np.median(salary):.1f}')
````

## 2. The owner joins the staff list

Add the owner's 200 and compare. Nobody else got a raise, yet the mean jumps by 15 while the median moves by 1. The median only looks at position in the sorted list.

````python
with_owner = np.append(salary, 200)

print(f'mean   {salary.mean():.1f} -> {with_owner.mean():.1f}')
print(f'median {np.median(salary):.1f} -> {np.median(with_owner):.1f}')
print(f'mean moved by {with_owner.mean() - salary.mean():.1f}, median by {np.median(with_owner) - np.median(salary):.1f}')
````

## 3. When the mean describes nobody

The lesson's small company: six salaries in thousands of dollars, with the founder at 800. The mean of 173.5 matches no employee. The median, 49.5, is a fair picture of a typical one.

````python
company = np.array([42, 45, 48, 51, 55, 800])

print(f'mean   = {company.mean():.1f}')
print(f'median = {np.median(company):.1f}')
````

## 4. The mode: the most common value

statistics.multimode returns every value tied for most frequent, so it also shows when there are two modes. It is the only one of the three that works for categories such as colours.

````python
shoes = [7, 8, 8, 9, 9, 9, 10]
ties = [1, 2, 2, 3, 3]
colours = ['red', 'blue', 'blue', 'green']

print('shoe sizes mode :', statistics.multimode(shoes))
print('tied example    :', statistics.multimode(ties))
print('colours mode    :', statistics.multimode(colours))
````

## The same idea in R, SPSS and Stata

**R**

```r
salary <- c(38, 42, 45, 47, 50, 52, 55, 58, 60)
mean(salary); median(salary)
with_owner <- c(salary, 200)

```

**SPSS** (Analyze > Descriptive Statistics > Frequencies > Statistics)

```
DATA LIST FREE / salary.
BEGIN DATA
38 42 45 47 50 52 55 58 60 200
END DATA.
FREQUENCIES VARIABLES=salary
  /FORMAT=NOTABLE
  /STATISTICS=MEAN MEDIAN MODE.
TEMPORARY.
SELECT IF (salary < 100).
FREQUENCIES VARIABLES=salary
  /FORMAT=NOTABLE
  /STATISTICS=MEAN MEDIAN MODE.
```

**Stata**

