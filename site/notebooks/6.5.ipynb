## 1. Enter the data

Nine nitrate readings from one river, in mg/L, and the guideline value we want to compare against. NumPy holds the numbers and SciPy supplies the test.

````python
import numpy as np
from scipy import stats

nitrate = np.array([3.1, 4.2, 2.8, 3.9, 4.5, 3.3, 4.1, 3.7, 3.5])
guideline = 3.2

print('readings:', len(nitrate))
````

## 2. Describe the sample

Always look at the numbers before testing them. The mean is the sample's centre, the standard deviation (ddof=1, the sample version) is its typical spread, and the standard error says how much a mean of 9 readings wobbles.

````python
mean = nitrate.mean()
sd = nitrate.std(ddof=1)
se = sd / np.sqrt(len(nitrate))

print(f'mean = {mean:.3f}')
print(f'sd   = {sd:.3f}')
print(f'se   = {se:.3f}')
````

## 3. Run the one-sample t-test

ttest_1samp asks: if the true average were exactly 3.2, how surprising is a sample mean this far away? It returns the t statistic (distance in standard errors) and the two-sided p-value.

````python
result = stats.ttest_1samp(nitrate, popmean=guideline)

print(f't  = {result.statistic:.3f}')
print(f'df = {len(nitrate) - 1}')
print(f'p  = {result.pvalue:.4f}')
````

## 4. Add a confidence interval

A 95% interval shows the range of true averages that fit the data. The guideline sits outside it, which agrees with the small p-value.

````python
ci = result.confidence_interval(confidence_level=0.95)

print(f'95% CI: ({ci.low:.3f}, {ci.high:.3f})')
print('guideline inside the interval?', ci.low <= guideline <= ci.high)
````
