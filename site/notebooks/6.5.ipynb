## The question

An environmental agency has a guideline of 3.2 mg/L for nitrate in river water. A technician takes nine water samples from one river and measures nitrate in each. Does the river's true average level differ from the guideline, or is the gap between the sample mean and 3.2 just the luck of which nine samples we took?

## Step 1: Enter the data

H0: the river's true mean nitrate is 3.2 mg/L. H1: it is not 3.2 (higher or lower). We have nine readings in mg/L and the guideline value. NumPy holds the numbers and SciPy supplies the test.

````python
import numpy as np  # Load NumPy, the library for lists of numbers, under the short name np.
from scipy import stats  # Load SciPy's statistics tools, which contain the t-test we need.

nitrate = np.array([3.1, 4.2, 2.8, 3.9, 4.5, 3.3, 4.1, 3.7, 3.5])  # Store the nine readings as an array so we can do arithmetic on all of them at once.
guideline = 3.2  # Store the guideline under a name, so the value we compare with is easy to find and change.

print('readings:', len(nitrate))  # Print the number of readings to confirm all nine were stored.
````

**What this means.** Nine readings are stored and the guideline is 3.2 mg/L. Nothing has been tested yet. The setup matters because the t-test needs the whole sample, not just its mean: the sample size and spread decide how much a mean can be trusted.

## Step 2: Describe the sample

Always look at the numbers before testing them. The mean is the sample's centre, the standard deviation (ddof=1, the sample version) is its typical spread, and the standard error says how much a mean of 9 readings wobbles.

````python
mean = nitrate.mean()  # The sample mean: the centre of the nine readings, our best guess for the river's true mean.
sd = nitrate.std(ddof=1)  # The sample SD (ddof=1 means divide by n - 1): how far a typical single reading is from the mean.
se = sd / np.sqrt(len(nitrate))  # The standard error: the SD divided by the square root of n. It says how much a MEAN of nine readings would wobble.

print(f'mean = {mean:.3f}')  # Print the mean to 3 decimals.
print(f'sd   = {sd:.3f}')  # Print the SD to 3 decimals.
print(f'se   = {se:.3f}')  # Print the standard error to 3 decimals.
````

**What this means.** The mean is 3.678, about 0.48 above the guideline. The SD is 0.554, so single readings typically vary by half a mg/L. The standard error is 0.185: a mean of nine readings wobbles far less than one reading does. The question becomes how many standard errors 3.678 is from 3.2.

## Step 3: Run the one-sample t-test

ttest_1samp asks: if the true average were exactly 3.2, how surprising is a sample mean this far away? It returns the t statistic (distance in standard errors) and the two-sided p-value.

````python
result = stats.ttest_1samp(nitrate, popmean=guideline)  # Run the one-sample t-test: is the mean of nitrate different from popmean = 3.2? It uses the t curve because we estimated the SD.

print(f't  = {result.statistic:.3f}')  # Print t, the distance between mean and guideline measured in standard errors.
print(f'df = {len(nitrate) - 1}')  # Print the degrees of freedom, n - 1 = 8, which sets the exact shape of the t curve.
print(f'p  = {result.pvalue:.4f}')  # Print the two-sided p-value to 4 decimals.
````

**What this means.** t = 2.587: our mean is about 2.6 standard errors above 3.2. With df = 8 the p-value is 0.0323. If the true mean were 3.2, a sample mean this far away (in either direction) would turn up about 3% of the time. That is below 0.05, so we reject H0. It is not the chance that H0 is true.

## Step 4: Add a confidence interval

A 95% interval shows the range of true averages that fit the data. The guideline sits outside it, which agrees with the small p-value.

````python
ci = result.confidence_interval(confidence_level=0.95)  # Ask the test result for a 95% confidence interval for the true mean river level.

print(f'95% CI: ({ci.low:.3f}, {ci.high:.3f})')  # Print the interval ends, rounded to 3 decimals.
print('guideline inside the interval?', ci.low <= guideline <= ci.high)  # Check whether the guideline lies inside the interval. This prints True or False.
````

**What this means.** The 95% interval for the river's true mean is 3.252 to 4.104. The guideline 3.2 is outside it (False), which agrees with p < 0.05. The interval also tells us how far: plausibly 0.05 to 0.9 mg/L above the guideline. It is wide because we have only nine readings.

## So, what is the answer?

The river's average nitrate level is estimated at 3.68 mg/L, and the one-sample t-test (t = 2.587, p = 0.0323) says it differs from the 3.2 guideline. The 95% interval, 3.25 to 4.10, lies entirely above 3.2. Caution: nine readings give a wide interval, the test assumes independent, roughly normal readings, and it cannot say why the level is high.

### Solution (try first, then peek)

````python
import numpy as np
from scipy import stats

grams = np.array([442.3, 421.3, 439.2, 443.6, 445.0, 430.2, 439.7, 435.2, 436.7, 453.5])
res = stats.ttest_1samp(grams, popmean=450)
ci = res.confidence_interval(0.95)
print(f'mean = {grams.mean():.2f}, sd = {grams.std(ddof=1):.2f}')
print(f't = {res.statistic:.3f}, df = {len(grams) - 1}, p = {res.pvalue:.5f}')
print(f'95% CI: ({ci.low:.2f}, {ci.high:.2f})')
````

**Answer.** The mean is 438.67 g and the SD is 8.74 g. t = -4.097 with df = 9 and p = 0.00269, far below 0.05, so we reject H0: the average loaf does not weigh 450 g. The 95% interval, 432.41 to 444.93 g, excludes 450, so loaves are plausibly 5 to 18 g light. With only 10 loaves from one bakery this is a sample, not a full audit.
