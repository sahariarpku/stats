## 1. Compare Maya's two exams

A z-score is (score - class mean) divided by class SD: how many standard deviations above or below average. Raw marks of 85 and 78 cannot be compared, but z-scores can.

````python
import numpy as np
from scipy import stats

z_stat = (85 - 70) / 10
z_bio = (78 - 65) / 6

print(f'Statistics: z = {z_stat:.2f}')
print(f'Biology:    z = {z_bio:.2f}')
print('Maya did better, relative to her class, in', 'Biology' if z_bio > z_stat else 'Statistics')
````

## 2. Standardise a whole data set

stats.zscore does every value at once. It divides by the population SD unless you set ddof=1, which gives the lesson's sample-SD values (-0.86, 0.26 ...).

````python
score = np.array([72, 85, 90, 68, 95])

z = stats.zscore(score, ddof=1)
print('z (ddof=1):', z.round(2))
print('z (default ddof=0):', stats.zscore(score).round(2))
````

## 3. Mean 0 and SD 1

Standardised data always have mean 0 and standard deviation 1. The shape and the ranking do not change, only the scale.

````python
print(f'mean of z = {abs(z.mean()):.2f}')
print(f'sd of z   = {z.std(ddof=1):.2f}')
print('same order as the scores?', (np.argsort(z) == np.argsort(score)).all())
````

## 4. Back to raw scores and percentiles

Reverse the formula with x = mean + z x SD. For bell-shaped data, norm.cdf turns a z-score into the share of values below it. Only trust that if the histogram is roughly bell-shaped.

````python
print('IQ with z = 2:', 100 + 2 * 15)
print(f'share below z = 1.5: {stats.norm.cdf(1.5):.3f}')
print(f'share below z = 2.17: {stats.norm.cdf(z_bio):.3f}')
````
