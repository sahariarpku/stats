## The question

Maya scored 85 in Statistics and 78 in Biology. The raw marks cannot be compared, because the classes differ: Statistics averages 70 with a standard deviation of 10, and Biology averages 65 with a standard deviation of 6. In which exam did she do better compared with her own class?

## Step 1: Compare Maya's two exams

A z-score is (score - class mean) divided by class SD: how many standard deviations above or below average. Raw marks of 85 and 78 cannot be compared, but z-scores can.

````python
import numpy as np  # Load NumPy, the library for lists of numbers, as np.
from scipy import stats  # Load SciPy's statistics tools; we use zscore and norm later.

z_stat = (85 - 70) / 10  # Statistics z-score: (her mark minus the class mean) divided by the class SD, so how many SDs above average she is.
z_bio = (78 - 65) / 6  # Biology z-score by the same recipe, with Biology's own mean (65) and SD (6).

print(f'Statistics: z = {z_stat:.2f}')  # Print the Statistics z-score, rounded to two decimals.
print(f'Biology:    z = {z_bio:.2f}')  # Print the Biology z-score.
print('Maya did better, relative to her class, in', 'Biology' if z_bio > z_stat else 'Statistics')  # Compare the two z-scores and print the subject where she stands further above her class.
````

**What this means.** The Statistics z-score is 1.50 (15 points above a class mean, with SD 10). The Biology z-score is 2.17 (13 points above, but the SD is only 6). So 78 in Biology is the stronger result, even though the raw mark is lower, because it sits 2.17 SDs above her class instead of 1.50.

## Step 2: Standardise a whole data set

stats.zscore does every value at once. It divides by the population SD unless you set ddof=1, which gives the lesson's sample-SD values (-0.86, 0.26 ...).

````python
score = np.array([72, 85, 90, 68, 95])  # Store the five quiz scores from the lesson in an array.

z = stats.zscore(score, ddof=1)  # stats.zscore standardises every value at once. ddof=1 divides by the sample SD, the version the lesson uses.
print('z (ddof=1):', z.round(2))  # Print the z-scores rounded to two decimals.
print('z (default ddof=0):', stats.zscore(score).round(2))  # Print them again with the default ddof=0 (population SD) to see the difference.
````

**What this means.** With the sample SD (11.6) the five scores become -0.86, 0.26, 0.69, -1.21 and 1.12. So the 95 is 1.12 SDs above the mean and the 68 is 1.21 below. The default ddof=0 divides by the smaller population SD (10.37), so each z-score is a little larger in size.

## Step 3: Mean 0 and SD 1

Standardised data always have mean 0 and standard deviation 1. The shape and the ranking do not change, only the scale.

````python
print(f'mean of z = {abs(z.mean()):.2f}')  # The mean of the z-scores. abs() avoids printing -0.00 caused by tiny rounding error.
print(f'sd of z   = {z.std(ddof=1):.2f}')  # The standard deviation of the z-scores, using the sample formula.
print('same order as the scores?', (np.argsort(z) == np.argsort(score)).all())  # Check the ranking: argsort gives the order of the values, and .all() checks every position matches.
````

**What this means.** The standardised scores have a mean of 0.00 and an SD of 1.00 by construction. The ranking is unchanged (True): z-scores only change the scale, not who scored higher or lower.

## Step 4: Back to raw scores and percentiles

Reverse the formula with x = mean + z x SD. For bell-shaped data, norm.cdf turns a z-score into the share of values below it. Only trust that if the histogram is roughly bell-shaped.

````python
print('IQ with z = 2:', 100 + 2 * 15)  # Go backwards with raw value = mean + z x SD. For IQ scores the mean is 100 and the SD 15.
print(f'share below z = 1.5: {stats.norm.cdf(1.5):.3f}')  # norm.cdf(1.5) gives the share of a bell curve lying below z = 1.5.
print(f'share below z = 2.17: {stats.norm.cdf(z_bio):.3f}')  # The same for Maya's Biology z-score, which turns it into a percentile.
````

**What this means.** A z of 2 on an IQ scale (mean 100, SD 15) is 130. If the marks are roughly bell-shaped, about 93.3% of students score below z = 1.5, and Maya's Biology z of 2.17 puts her above about 98.5% of her class. Those percentages are only trustworthy when the data are close to bell-shaped.

## So, what is the answer?

Measured against her own classes, Maya did better in Biology (z = 2.17) than in Statistics (z = 1.50), even though the raw Biology mark is lower. A z-score counts how many SDs a value sits from its mean, puts different scales on one yardstick and keeps the ranking. Caution: turning a z-score into a percentage (98.5% for 2.17) is only trustworthy when the marks are roughly bell-shaped.

## Now you try

Dana took an entrance test with two parts. She scored 66 in Maths, where applicants average 52 with SD 8, and 71 in Writing, where they average 60 with SD 11. She also kept a log of her study hours over six weeks and wants them standardised.

**Data:** hours = [4, 7, 5, 9, 6, 12]; maths: score 66, mean 52, sd 8; writing: score 71, mean 60, sd 11

1. Compute Dana's z-score in Maths and in Writing, and say which part is stronger.
2. Standardise the six study hours (sample SD).
3. Assuming a bell shape, find the share of applicants below her Maths z-score.

*Hint:* z = (score - mean) / sd. For the list use stats.zscore with ddof=1, and stats.norm.cdf turns a z-score into a share.

````python
# Your turn: write your code here
````

### Solution (try first, then peek)

````python
import numpy as np
from scipy import stats

z_math = (66 - 52) / 8
z_writing = (71 - 60) / 11
print(f'maths z = {z_math:.2f}, writing z = {z_writing:.2f}')
hours = np.array([4, 7, 5, 9, 6, 12])
z = stats.zscore(hours, ddof=1)
print('z of the six values:', z.round(2))
print(f'share below the maths z (bell shape assumed): {stats.norm.cdf(z_math):.3f}')
````

**Answer.** Maths z = 1.75 and Writing z = 1.00, so Maths is the stronger part, even though 71 is the higher raw mark. The study hours become -1.08, -0.06, -0.74, 0.63, -0.40 and 1.65, so the 12-hour week is the most unusual. Assuming a bell shape, about 96.0% of applicants scored below her Maths mark.
