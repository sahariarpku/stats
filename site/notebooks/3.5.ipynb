## 1. Check the condition

The normal curve is a good stand-in for the binomial only when both np and n(1 - p) are at least 10. Here n = 200 and p = 0.1, so we check both numbers first.

````python
n, p = 200, 0.1

print('np      =', n * p)
print('n(1-p)  =', n * (1 - p))
print('safe to approximate?', n * p >= 10 and n * (1 - p) >= 10)
````

## 2. The exact binomial answer

Fewer than 15 no-shows means 14 or fewer. The binomial cdf adds up all fifteen terms for us, so this is the true value we want the normal curve to match.

````python
from scipy import stats

exact = stats.binom.cdf(14, n, p)

print(f'exact P(X <= 14) = {exact:.4f}')
````

## 3. The normal approximation

Match the normal curve to the binomial's mean np and SD sqrt(np(1 - p)). The continuity correction uses 14.5 because the bar for 14 stretches from 13.5 to 14.5.

````python
from math import sqrt

mu = n * p
sigma = sqrt(n * p * (1 - p))

with_cc = stats.norm.cdf(14.5, mu, sigma)
without = stats.norm.cdf(14, mu, sigma)

print(f'mu = {mu:.1f}, sigma = {sigma:.3f}')
print(f'normal with correction    = {with_cc:.4f}')
print(f'normal without correction = {without:.4f}')
````

## 4. How close is it?

The error with the half-unit correction is far smaller than without it. That is why the correction matters, even for a big n.

````python
print(f'error with correction    = {abs(with_cc - exact):.4f}')
print(f'error without correction = {abs(without - exact):.4f}')
````

## The same idea in R, SPSS and Stata

**R**

```r
n <- 200; p <- 0.1
pbinom(14, n, p)
pnorm(14.5, mean = n * p, sd = sqrt(n * p * (1 - p)))
pnorm(14,   mean = n * p, sd = sqrt(n * p * (1 - p)))
```

**SPSS** (Transform > Compute Variable (use CDF.BINOM and CDF.NORMAL))

```
DATA LIST FREE / dummy.
BEGIN DATA
1
END DATA.
COMPUTE exact = CDF.BINOM(14, 200, 0.1).
COMPUTE mu = 200 * 0.1.
COMPUTE sigma = SQRT(200 * 0.1 * 0.9).
COMPUTE with_cc = CDF.NORMAL(14.5, mu, sigma).
COMPUTE without_cc = CDF.NORMAL(14, mu, sigma).
FORMATS exact with_cc without_cc (F8.4).
LIST exact with_cc without_cc.
```
