## The question

An airline sells 200 tickets on a flight and about 10% of ticket holders do not show up. The manager asks: what is the chance that fewer than 15 passengers fail to show? The exact answer needs the binomial distribution, but a normal curve is much easier to use. How close does the normal approximation get, and does the half-unit continuity correction help?

## Step 1: Check the condition

The normal curve is a good stand-in for the binomial only when both np and n(1 - p) are at least 10. Here n = 200 and p = 0.1, so we check both numbers first.

````python
n, p = 200, 0.1  # Store the number of tickets and the no-show probability.

print('np      =', n * p)  # Print n x p, the expected number of no-shows.
print('n(1-p)  =', n * (1 - p))  # Print n x (1 - p), the expected number of shows.
print('safe to approximate?', n * p >= 10 and n * (1 - p) >= 10)  # Check the rule of thumb: both must be at least 10 for the normal curve to be a good stand-in.
````

**What this means.** The expected no-shows are 20 and the expected shows 180, both well above 10, so the check prints True. The normal approximation is safe to use here.

## Step 2: The exact binomial answer

Fewer than 15 no-shows means 14 or fewer. The binomial cdf adds up all fifteen terms for us, so this is the true value we want the normal curve to match.

````python
from scipy import stats  # Load the statistics tools from SciPy.

exact = stats.binom.cdf(14, n, p)  # Fewer than 15 means 14 or fewer, so ask the binomial cdf at 14. This is the exact answer.

print(f'exact P(X <= 14) = {exact:.4f}')  # Print the exact probability.
````

**What this means.** The exact chance of fewer than 15 no-shows is 0.0929, about 9.3%. This is the true value our approximation must match.

## Step 3: The normal approximation

Match the normal curve to the binomial's mean np and SD sqrt(np(1 - p)). The continuity correction uses 14.5 because the bar for 14 stretches from 13.5 to 14.5.

````python
from math import sqrt  # Load the square root function.

mu = n * p  # The matching normal curve has the binomial's mean n x p.
sigma = sqrt(n * p * (1 - p))  # It has the binomial's SD, the square root of n x p x (1 - p).

with_cc = stats.norm.cdf(14.5, mu, sigma)  # Apply the continuity correction: use 14.5, the edge of the bar for 14, when using the normal.
without = stats.norm.cdf(14, mu, sigma)  # Without the correction: use 14, the middle of the bar.

print(f'mu = {mu:.1f}, sigma = {sigma:.3f}')  # Print the mean and SD of the matching normal.
print(f'normal with correction    = {with_cc:.4f}')  # Print the answer with the correction.
print(f'normal without correction = {without:.4f}')  # Print the answer without the correction.
````

**What this means.** The matching normal has mean 20.0 and SD 4.243. With the correction it gives 0.0974, without it 0.0786, compared with the exact 0.0929. The corrected value is clearly the closer.

## Step 4: How close is it?

The error with the half-unit correction is far smaller than without it. That is why the correction matters, even for a big n.

````python
print(f'error with correction    = {abs(with_cc - exact):.4f}')  # Print how far the corrected answer is from the exact one.
print(f'error without correction = {abs(without - exact):.4f}')  # Print how far the uncorrected answer is.
````

**What this means.** The error with the correction is only 0.0045, but without it the error is 0.0143, about three times bigger. The half-unit shift is worth doing even with a big n.

## So, what is the answer?

The chance that fewer than 15 passengers fail to show is 0.0929 exactly. The normal approximation gives 0.0974 with the continuity correction and 0.0786 without it, so the correction cuts the error from 0.0143 to 0.0045. The caution: the approximation is only trustworthy when np and n(1 - p) are both at least 10.

### Solution (try first, then peek)

````python
from math import sqrt
from scipy import stats

n, p = 500, 0.4
mu, sigma = n * p, sqrt(n * p * (1 - p))
print(f'np = {mu}, n(1-p) = {n * (1 - p)}, sigma = {sigma:.3f}')

exact = stats.binom.cdf(185, n, p)
with_cc = stats.norm.cdf(185.5, mu, sigma)
without = stats.norm.cdf(185, mu, sigma)
print(f'exact = {exact:.4f}, with correction = {with_cc:.4f}, without = {without:.4f}')
print(f'errors: {abs(with_cc - exact):.4f} and {abs(without - exact):.4f}')
````

**Answer.** np = 200 and n(1 - p) = 300, both far above 10, and the matching SD is 10.954. The exact P(X <= 185) is 0.0924. The normal gives 0.0928 with the correction and 0.0855 without. The errors are 0.0004 and 0.0069, so the corrected version is much closer.

**SPSS** (Transform > Compute Variable (use CDF.BINOM and CDF.NORMAL))

```
DATA LIST FREE / dummy.
BEGIN DATA
1
END DATA.
COMPUTE exact = CDF.BINOM(14, 200, 0.1).
COMPUTE mu = 200 * 0.1.

FORMATS exact with_cc without_cc (F8.4).
LIST exact with_cc without_cc.
```

How to read it: LIST shows exact = 0.0929, with_cc = 0.0974 and without_cc = 0.0786. The corrected normal is closer to the exact answer, as in Python.
