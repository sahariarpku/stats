## The question

A maths teacher has the end-of-term scores of her 30 students on her desk. She has two questions: how did my class do, and what can I say about similar students I never tested? We have the 30 scores, and we will answer the first question by describing and the second by inferring.

## Step 1: Enter the 30 test scores

The teacher's 30 end-of-term scores, held in a NumPy array. The mean is 74 and the standard deviation about 12.3, as in the lesson.

````python
import numpy as np  # Load NumPy, the library for lists of numbers, under the short name np so we can type np. instead of numpy.
from scipy import stats  # Load SciPy's statistics tools; we need them in step 3 for the confidence interval.

scores = np.array([41,57,61,62,62,62,65,65,68,68,70,72,72,73,73,75,75,76,77,77,78,79,80,84,86,86,88,95,95,98])  # Store the 30 scores in an array, so one command can work on all of them at once.

print('students:', len(scores))  # Print how many scores we have. Always check n first, because everything later depends on it.
print('lowest to highest:', scores.min(), 'to', scores.max())  # Print the lowest and highest score, to see how far the results stretch before we compute anything.
````

**What this means.** 30 students, with scores from 41 to 98. That is a wide spread, so one average will hide differences between students. This is why we measure the spread as well as the centre in the next step.

## Step 2: Describe: summarise what we have

Descriptive statistics only talk about these 30 students. The mean and median give the centre, and the standard deviation (ddof=1, the sample version) gives the typical distance from the average.

````python
mean = scores.mean()  # The mean (add all scores, divide by 30) is the balance point of the class and our headline answer to 'how did we do?'.
median = np.median(scores)  # The median is the middle score after sorting. We compute it too, to see whether a few extreme scores are pulling the mean.
sd = scores.std(ddof=1)  # Standard deviation, the typical distance of a score from the mean. ddof=1 divides by n - 1, because a sample under-estimates the spread of its whole group.

print(f'mean   = {mean:.1f}')  # Print the mean. The f-string puts the number inside the text, and :.1f rounds it to one decimal place.
print(f'median = {median:.1f}')  # Print the median on the next line so we can compare it with the mean.
print(f'sd     = {sd:.1f}')  # Print the standard deviation, the number that says how far scores typically sit from the average.
print(f'about 2/3 of scores lie within {mean - sd:.1f} to {mean + sd:.1f}')  # Mean plus and minus one SD marks the range where about two thirds of bell-shaped scores fall: a picture of a typical student.
````

**What this means.** The mean and the median are both 74.0, so the class is balanced and no extreme scores drag the average. The SD of 12.3 means a typical student sits about 12 points from 74, and about two thirds of the scores fall between 61.7 and 86.3. These are plain facts about these 30 students.

## Step 3: Infer: a 95% confidence interval

Now treat the 30 students as a sample from a bigger group. stats.t.interval gives a range of plausible values for that group's true mean. The standard error (sd divided by the square root of n) says how much a mean of 30 scores wobbles.

````python
n = len(scores)  # Store n, the number of scores. len() counts them, and we need n for both the standard error and the degrees of freedom.
se = sd / np.sqrt(n)  # Standard error = sd / square root of n. It says how much a mean of 30 scores would wobble from one sample to the next.

low, high = stats.t.interval(0.95, df=n - 1, loc=mean, scale=se)  # stats.t.interval builds a 95% range for the true mean. It uses the t distribution because we estimated the spread from the sample, with df = n - 1.

print(f'standard error = {se:.2f}')  # Print the standard error to two decimals, so we can see how precise our mean is.
print(f'95% CI for the true mean: ({low:.1f}, {high:.1f})')  # Print the two ends of the interval: the low and high values of the true mean that fit our data.
````

**What this means.** The standard error is 2.24, so a mean of 30 scores like these would typically wobble by about 2 points. The 95% interval is 69.4 to 78.6. If these students are a fair sample of a wider group, that group's true average plausibly lies in this range. The 74.0 is only our best single guess.

## Step 4: What each one tells you

The describing numbers are exact facts about these 30 scores. The interval is an estimate about people we did not test, and it comes with stated uncertainty. That is the whole difference between the two branches.

````python
print('DESCRIBE: this class averaged', round(mean, 1), 'with a typical spread of', round(sd, 1))  # Print the descriptive statement. It is a fact about these 30 students only, so it carries no uncertainty.
print('INFER: a wider group of similar students probably averages between', round(low, 1), 'and', round(high, 1))  # Print the inferential statement. It is about students we did not test, so we give a range instead of one number.
print('Is the interval narrower than the spread of scores?', (high - low) < 2 * sd)  # Compare the interval's width with 2 SDs. The interval is much narrower, because it locates the average, not individual students.
````

**What this means.** True: the interval is about 9 points wide, far narrower than the roughly 25 points over which individual scores spread. Do not mix them up. The standard deviation describes how students differ, while the interval says where the group's average probably lies.

## So, what is the answer?

The class averaged 74 with a typical spread of about 12 points: that is a plain fact about these 30 students. If they are a fair sample of similar students, the true average of that wider group plausibly lies between 69.4 and 78.6. Caution: that range is only trustworthy if the 30 students are representative of the group we want to talk about.

## Now you try

A nurse measures the resting pulse (beats per minute) of 20 volunteers at a health fair. She wants to describe these 20 people, and also to estimate the average resting pulse of all adults who might come to such fairs.

**Data:** pulse = [64, 70, 72, 68, 75, 80, 66, 71, 74, 69, 77, 73, 62, 78, 72, 67, 70, 76, 65, 74]

1. Find the mean, median and sample standard deviation of the pulses.
2. Compute the standard error and a 95% confidence interval for the true mean.
3. Say which results describe the 20 volunteers and which one is an inference about all fair visitors.

*Hint:* Follow the lab's two branches: describe with mean, median and std(ddof=1), then infer with the standard error and stats.t.interval.

### Solution (try first, then peek)

````python
import numpy as np
from scipy import stats

pulse = np.array([64, 70, 72, 68, 75, 80, 66, 71, 74, 69, 77, 73, 62, 78, 72, 67, 70, 76, 65, 74])
n = len(pulse)
mean, sd = pulse.mean(), pulse.std(ddof=1)
se = sd / np.sqrt(n)
low, high = stats.t.interval(0.95, df=n - 1, loc=mean, scale=se)
print(f'n = {n}, mean = {mean:.1f}, median = {np.median(pulse):.1f}, sd = {sd:.2f}')
print(f'standard error = {se:.2f}')
print(f'95% CI: ({low:.1f}, {high:.1f})')
````

**Answer.** Mean 71.2, median 71.5 and sd 4.88: the volunteers sit about 5 beats from 71. The standard error is 1.09 and the 95% interval is 68.9 to 73.4. The first three numbers describe the 20 volunteers, while the interval infers the average of all fair visitors, assuming the volunteers are typical of them.
