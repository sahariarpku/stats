## 1. Enter the 30 test scores

The teacher's 30 end-of-term scores, held in a NumPy array. The mean is 74 and the standard deviation about 12.3, as in the lesson.

````python
import numpy as np
from scipy import stats

scores = np.array([41,57,61,62,62,62,65,65,68,68,70,72,72,73,73,75,75,76,77,77,78,79,80,84,86,86,88,95,95,98])

print('students:', len(scores))
print('lowest to highest:', scores.min(), 'to', scores.max())
````

## 2. Describe: summarise what we have

Descriptive statistics only talk about these 30 students. The mean and median give the centre, and the standard deviation (ddof=1, the sample version) gives the typical distance from the average.

````python
mean = scores.mean()
median = np.median(scores)
sd = scores.std(ddof=1)

print(f'mean   = {mean:.1f}')
print(f'median = {median:.1f}')
print(f'sd     = {sd:.1f}')
print(f'about 2/3 of scores lie within {mean - sd:.1f} to {mean + sd:.1f}')
````

## 3. Infer: a 95% confidence interval

Now treat the 30 students as a sample from a bigger group. stats.t.interval gives a range of plausible values for that group's true mean. The standard error (sd divided by the square root of n) says how much a mean of 30 scores wobbles.

````python
n = len(scores)
se = sd / np.sqrt(n)

low, high = stats.t.interval(0.95, df=n - 1, loc=mean, scale=se)

print(f'standard error = {se:.2f}')
print(f'95% CI for the true mean: ({low:.1f}, {high:.1f})')
````

## 4. What each one tells you

The describing numbers are exact facts about these 30 scores. The interval is an estimate about people we did not test, and it comes with stated uncertainty. That is the whole difference between the two branches.

````python
print('DESCRIBE: this class averaged', round(mean, 1), 'with a typical spread of', round(sd, 1))
print('INFER: a wider group of similar students probably averages between', round(low, 1), 'and', round(high, 1))
print('Is the interval narrower than the spread of scores?', (high - low) < 2 * sd)
````
