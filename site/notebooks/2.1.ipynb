## 1. List the sample space of two coins

itertools.product writes out every sequence of two flips, just like the lesson's table. Counting the sequences with exactly one head gives the exact probability, with no guessing.

````python
from itertools import product

flips = list(product('HT', repeat=2))
one_head = [f for f in flips if f.count('H') == 1]

print('sample space:', flips)
print('exactly one head:', one_head)
print(f'P(exactly one head) = {len(one_head)}/{len(flips)} = {len(one_head) / len(flips):.4f}')
````

## 2. Count the ways to roll a 7

Two dice give 6 x 6 = 36 equally likely pairs. We count the pairs that add to 7 and divide by 36. This exact value is the target the simulation should approach.

````python
pairs = list(product(range(1, 7), repeat=2))
sevens = [p for p in pairs if sum(p) == 7]
exact = len(sevens) / len(pairs)

print('pairs:', len(pairs))
print('pairs that sum to 7:', len(sevens))
print(f'exact P(sum = 7) = {exact:.4f}')
````

## 3. Simulate rolling two dice

A seeded generator rolls two dice again and again, so the result is the same every time you run it. The relative frequency is (times a 7 came up) divided by (rolls). Watch it settle near the exact value.

````python
import numpy as np

rng = np.random.default_rng(42)

for n in [10, 100, 10_000, 1_000_000]:
    total = rng.integers(1, 7, size=n) + rng.integers(1, 7, size=n)
    freq = (total == 7).mean()
    print(f'{n:>9,} rolls: relative frequency = {freq:.4f}')

print(f'exact value              = {exact:.4f}')
````
