## The question

## Step 1: List the sample space of two coins

itertools.product writes out every sequence of two flips, just like the lesson's table. Counting the sequences with exactly one head gives the exact probability, with no guessing.

````python
from itertools import product  # Load product from itertools, a standard tool that lists every combination of choices, so we do not have to write them by hand.

flips = list(product('HT', repeat=2))  # Write out every sequence of two flips (HH, HT, TH, TT). This list is the sample space, the complete set of outcomes.
one_head = [f for f in flips if f.count('H') == 1]  # Keep only the sequences with exactly one H. These are the outcomes that count as our event.

print('sample space:', flips)  # Print the full sample space so you can see all four equally likely outcomes with your own eyes.
print('exactly one head:', one_head)  # Print the favourable outcomes so you can check that HT and TH are the two we expected.
print(f'P(exactly one head) = {len(one_head)}/{len(flips)} = {len(one_head) / len(flips):.4f}')  # Divide the favourable count by the total count: probability is favourable over possible, when outcomes are equally likely.
````

**What this means.** There are 4 equally likely sequences and 2 of them have exactly one head (HT and TH), so P = 2/4 = 0.5. Notice that the order matters when we list outcomes: HT and TH are different sequences, and that is why the answer is 1/2 and not 1/3.

## Step 2: Count the ways to roll a 7

Two dice give 6 x 6 = 36 equally likely pairs. We count the pairs that add to 7 and divide by 36. This exact value is the target the simulation should approach.

````python
pairs = list(product(range(1, 7), repeat=2))  # Make all 36 (die 1, die 2) pairs. Listing them is how we make sure no outcome is forgotten.
sevens = [p for p in pairs if sum(p) == 7]  # Keep the pairs whose two numbers add to 7. These are the outcomes we want.
exact = len(sevens) / len(pairs)  # Divide the number of winning pairs by the number of all pairs. Store it as exact, because we will compare with it later.

print('pairs:', len(pairs))  # Print how many pairs there are, to confirm we really have 6 x 6 = 36.
print('pairs that sum to 7:', len(sevens))  # Print how many of them sum to 7, the top of our fraction.
print(f'exact P(sum = 7) = {exact:.4f}')  # Print the exact probability, rounded to four decimals so it is easy to read.
````

**What this means.** Of the 36 equally likely pairs, 6 add up to 7, so the exact probability is 6/36 = 0.1667, about 1 roll in 6. A 7 is the most common total because it has more ways to happen than any other sum. This exact value is the target for the simulation next.

## Step 3: Simulate rolling two dice

A seeded generator rolls two dice again and again, so the result is the same every time you run it. The relative frequency is (times a 7 came up) divided by (rolls). Watch it settle near the exact value.

````python
import numpy as np  # Load NumPy, the library that gives us fast random numbers.

rng = np.random.default_rng(42)  # Create a random number generator with seed 42. Fixing the seed means everyone gets the same rolls and the results can be compared.

for n in [10, 100, 10_000, 1_000_000]:  # Loop over four sample sizes, from a handful of rolls to a million, to watch what happens as we roll more.
    total = rng.integers(1, 7, size=n) + rng.integers(1, 7, size=n)  # Roll two dice n times (integers 1 to 6 from each) and add them, one total per roll.
    freq = (total == 7).mean()  # Count the share of totals that equal 7. True counts as 1, so the mean of the True/False list is the relative frequency.
    print(f'{n:>9,} rolls: relative frequency = {freq:.4f}')  # Print each frequency next to its number of rolls, with thousands separators so big numbers stay readable.

print(f'exact value              = {exact:.4f}')  # Print the exact value once more so the simulated lines can be compared with it.
````

**What this means.** With 10 rolls the frequency was 0.0000 (no 7 showed up at all), with 100 rolls 0.1700, with 10,000 rolls 0.1663 and with a million rolls 0.1669. The exact value is 0.1667. Few rolls give wild answers; many rolls settle close to the exact probability.

## So, what is the answer?

Exactly one head from two coins has probability 0.5, and a total of 7 from two dice has probability 6/36 = 0.1667. The simulation agreed once we rolled many times (0.1669 for a million rolls). The caution: this counting works only because every outcome is equally likely, and a small simulation can miss the exact value by a lot.

A board game uses two four-sided dice (faces 1 to 4). A player needs a total of exactly 5 to move into the castle. Before playing, the designer wants to know how likely that is, and to confirm the answer with a computer simulation.

**Data:** Each die has faces [1, 2, 3, 4]. Use range(1, 5) to list them. For the simulation use the seed 7 and 20, 1000 and 100000 rolls.

1. List all pairs of faces and count how many are possible.
2. How many of them total 5, and what is the exact probability?
3. Simulate 20, 1,000 and 100,000 double rolls with seed 7. Which frequency is closest to the exact value, and why?

*Hint:* Use product(range(1, 5), repeat=2) for the pairs, then divide the winning pairs by all pairs. In the simulation, integers(1, 5) gives faces 1 to 4.

### Solution (try first, then peek)

````python
from itertools import product
import numpy as np

pairs = list(product(range(1, 5), repeat=2))
hits = [p for p in pairs if sum(p) == 5]
exact = len(hits) / len(pairs)
print('pairs:', len(pairs), ' sum 5:', len(hits), f' exact = {exact:.4f}')

rng = np.random.default_rng(7)
for n in [20, 1000, 100_000]:
    total = rng.integers(1, 5, size=n) + rng.integers(1, 5, size=n)
    print(f'{n:>7,} rolls: {(total == 5).mean():.4f}')
````

**Answer.** There are 16 pairs and 4 of them total 5 ((1,4), (2,3), (3,2), (4,1)), so the exact probability is 4/16 = 0.25. The simulated frequencies are 0.2500 for 20 rolls, 0.2420 for 1,000 and 0.2477 for 100,000. The big run is closest in general because more rolls average out the luck; the 20-roll value matched only by chance.

```r

mean(pairs == 7)                  # exact P(sum = 7)  # The share of the 36 sums equal to 7 is the exact probability, because a True/False test averages to a proportion.

flips <- expand.grid(c("H", "T"), c("H", "T"))  # List all coin pairs with expand.grid(), R's way of crossing every choice with every other.
heads <- (flips$Var1 == "H") + (flips$Var2 == "H")  # Count the heads in each pair by adding two True/False tests together.
mean(heads == 1)                  # exact P(exactly one head)  # The share of pairs with exactly 1 head is the exact probability for the coins.

set.seed(42)  # Fix the seed so the random numbers are repeatable.

mean(sample(6, n, TRUE) + sample(6, n, TRUE) == 7)  # Roll two dice 10,000 times, add them, and take the share of 7s.
```

How to read it: Look at the three numbers printed in order: 0.1667 for a sum of 7, 0.5 for exactly one head, and a simulated value near 0.16 to 0.17. The first two match the counting in Python exactly. The third is random, so it is close to 0.1667 but not identical.
